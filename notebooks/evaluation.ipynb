# Import necessary librairies and load data

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, roc_curve, confusion_matrix, classification_report
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import numpy as np
```

```python
# Chargement des données de test (data 1)
X_test = np.load('../datasets/dataset1/X_test_features.npy')
y_test = np.load('../datasets/dataset1/y_test.npy')
```

```python
# Chargement des données de test (data 2)
X_test = np.load('../datasets/dataset2/X_test_features.npy')
y_test = np.load('../datasets/dataset2/y_test.npy')
```

```python
# Chargement des données de test (data 3)
X_test = np.load('../datasets/dataset3/X_test_features.npy')
y_test = np.load('../datasets/dataset3/y_test.npy')
```

```python
# Chargement des données de test (data 4)
X_test = np.load('../datasets/dataset4/X_test_features.npy')
y_test = np.load('../datasets/dataset4/y_test.npy')
```

```python
# Chargement des données de test (data 5)
X_test = np.load('../datasets/dataset5/X_test_features.npy')
y_test = np.load('../datasets/dataset5/y_test.npy')
```

# SVM

```python
# Chargement du modèle entraîné
model_path = '../models/best_svm_model.pkl'
svm = joblib.load(model_path)
```

```python
# Predict the test set
y_pred = svm.predict(X_test)
y_proba = svm.predict_proba(X_test)[:, 1]
```

```python
# Compute evaluation metrics
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
roc_auc = roc_auc_score(y_test, y_proba)

# Print the evaluation metrics
print(f'Accuracy: {accuracy:.4f}')
print(f'Precision: {precision:.4f}')
print(f'Recall: {recall:.4f}')
print(f'F1 Score: {f1:.4f}')
print(f'ROC AUC Score: {roc_auc:.4f}')
```

```python
# Matrice de confusion
conf_matrix = confusion_matrix(y_test, y_pred)
print('Confusion Matrix:')
print(conf_matrix)
```

```python
# Confusion Matrix
conf_matrix = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(10, 7))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues')
plt.title('Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```

```python
# Classification Report
print('Classification Report:')
print(classification_report(y_test, y_pred))
```

```python
# ROC Curve
fpr, tpr, _ = roc_curve(y_test, y_proba)
plt.figure(figsize=(10, 7))
plt.plot(fpr, tpr, color='blue', label=f'ROC curve (area = {roc_auc:.4f})')
plt.plot([0, 1], [0, 1], color='grey', linestyle='--')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Receiver Operating Characteristic (ROC) Curve')
plt.legend()
plt.show()
```
